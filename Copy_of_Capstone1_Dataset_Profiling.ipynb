```python
from scipy.stats import spearmanr

# Calculate consultations per service day
staffing_analysis = phcc_integrated.copy()

staffing_analysis["Consultations_Per_Service_Day"] = (
    staffing_analysis["Total_Consultations"] /
    staffing_analysis["Service_Days"]
)

# Keep records with both variables available
staffing_analysis = staffing_analysis.dropna(
    subset=[
        "Staff_Availability_Pct_Corrected",
        "Consultations_Per_Service_Day"
    ]
)

# Calculate Spearman correlation
correlation, p_value = spearmanr(
    staffing_analysis["Staff_Availability_Pct_Corrected"],
    staffing_analysis["Consultations_Per_Service_Day"]
)

print("Spearman Correlation:", round(correlation,3))
print("p-value:", round(p_value,4))
print("Observations:",len(staffing_analysis))

# Scatter plot
plt.figure(figsize=(9,6))

plt.scatter(
    staffing_analysis["Staff_Availability_Pct_Corrected"],
    staffing_analysis["Consultations_Per_Service_Day"]
)

plt.title("Staff Availability vs. Consultations per Service Day")
plt.xlabel("Staff Availability (%)")
plt.ylabel("Consultations per Service Day")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import kruskal

fully_functional = phcc_integrated.loc[
    phcc_integrated["Functionality_Status"] == "Fully functional",
    "Consultations_Per_Service_Day"
].dropna()

partially_functional = phcc_integrated.loc[
    phcc_integrated["Functionality_Status"] == "Partially functional",
    "Consultations_Per_Service_Day"
].dropna()

temporarily_disrupted = phcc_integrated.loc[
    phcc_integrated["Functionality_Status"] == "Temporarily disrupted",
    "Consultations_Per_Service_Day"
].dropna()

statistic, p_value = kruskal(
    fully_functional,
    partially_functional,
    temporarily_disrupted
)

print("Kruskal-Wallis Statistic:", round(statistic,3))
print("p-value:", round(p_value,4))
```

```python
resource_distribution = (
    phcc_integrated
    .groupby("Facility_ID")
    .agg(
        Total_Allocated_EUR = ("Total_Allocated_EUR","sum"),
        Total_Spent_EUR = (
            "Total_Spent_EUR","sum"
        ),
        Months = ("Month","nunique")
    )
    .reset_index()
)

resource_distribution["Average_Monthly_Allocation_EUR"] = (
    resource_distribution["Total_Allocated_EUR"] /
    resource_distribution["Months"]
)

resource_distribution["Average_Monthly_Expenditure_EUR"] = (
    resource_distribution["Total_Spent_EUR"] /
    resource_distribution["Months"]
)

resource_distribution["Budget_Utilisation_Pct"] = (
    resource_distribution["Total_Spent_EUR"] /
    resource_distribution["Total_Allocated_EUR"]
) * 100

display(
    resource_distribution.sort_values(
        "Total_Allocated_EUR",
        ascending=False
    ).round(2)
)
```
